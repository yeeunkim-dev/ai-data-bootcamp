# numpy 숫자 덩어리를 한 번에

```python
# 스스로 해보자! (3)
scores = np.array([72, 88, 95, 60, 78, 91, 55, 83])

# 아래 주석(#)을 지우고 빈칸을 채워보세요.
print("평균:", scores.mean())
print("표준편차:", scores.std().round(1))
above = scores > scores.mean() # 2) 평균보다 높은가? (불리언 배열)
print(above.sum(), "명")       # 3) 이 값의 의미는 무엇일까요? 평균보다 높은 사람의 비율

# scores[above]        # 평균보다 높은 점수만 골라내기
# above.sum()          # 평균보다 높은 사람 수 (True = 1로 계산)
# above.mean()         # 평균보다 높은 사람의 비율
```

# 차트 유형 선택과 시각화 - 목적이 차트를 결정한다

```python
# 스스로 해보자! (6)
# 1) 나이 분포 (이상치 포함)
sns.histplot(customers["age"]); plt.title("Age (with outliers)"); plt.show()

# 2) 이상치 제거 후
mask = (customers["age"] > 0) & (customers["age"] < 120)
sns.histplot(customers.loc[mask, "age"]); plt.title("Age (cleaned)"); plt.show()
```
